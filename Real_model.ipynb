```python
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.feature_selection import mutual_info_classif
import matplotlib.pyplot as plt
import pandas as pd
```

```python
data = load_breast_cancer()
```

```python
table = pd.DataFrame(data.data,columns=data.feature_names)
table.head()
```

```python
table.shape
```

```python
data.target
```

```python
imp = mutual_info_classif(data.data, data.target)
fimp = pd.Series(imp,table.columns[0:len(table.columns)])
fimp.sort_values(ascending=False)
```

```python
final=fimp.index[0:10]
final
```

```python
x = table[final]
y = data.target
```

```python
x
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
scaled = scaler.fit_transform(x)

from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(scaled, y, test_size = 0.25, random_state=0)
```

```python
from sklearn.ensemble import RandomForestClassifier
m1 = RandomForestClassifier(n_estimators=100)
m1.fit(X_train,y_train)
```

```python
m1.score(X_test, y_test)
```

```python
import pickle

pickle.dump(m1, open("m1.pkl", "wb"))
pickle.dump(scaler, open("scaler.pkl", "wb"))
pickle.dump(list(final), open("features.pkl", "wb"))
```
